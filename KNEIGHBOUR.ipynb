```python
import pandas as pd
import numpy as np
import seaborn as sns 
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from imblearn.over_sampling import SMOTE
```

```python
df=pd.read_csv(r"C:\Users\ADMIN\Downloads\customers (3).csv")
```

```python
x=df.drop("Channel",axis=1)
x
```

```python
y=df["Channel"]
y
```

```python
sm=SMOTE(random_state=42,k_neighbors=5)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.80,random_state=42,stratify=y)
```

```python
x_train,y_train=sm.fit_resample(x_train,y_train)
```

```python
df.describe()
```

```python
x.columns
```

```python
ct=ColumnTransformer(transformers=[
    ("tnf1",StandardScaler(),['Region', 'Fresh', 'Milk', 'Grocery', 'Frozen', 'Detergents_Paper',
       'Delicatessen'])
],remainder="passthrough",)
```

```python
pipe=Pipeline([("preprocessor",ct),("model",KNeighborsClassifier(n_neighbors=3,))])
```

```python
pipe.fit(x_train,y_train)
```

```python
y_pre=pipe.predict(x_test)
```

```python
y_pre
```

```python
cm=confusion_matrix(y_test,y_pre)
```

```python
cm
```

```python
cr=classification_report(y_test,y_pre)
```

```python
cr
```

```python
accuracy_score(y_test,y_pre)
```
