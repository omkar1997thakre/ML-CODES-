```python
x=df.drop("Channel",axis=1)
x
```

```python
y=df["Channel"]
y
```

```python
x.columns
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.80,random_state=42,stratify=y)
```

```python
df["Channel"].value_counts()
```

```python
sm=SMOTE(random_state=42,k_neighbors=5)
```

```python
x_train,y_train=sm.fit_resample(x_train,y_train)
```

```python
cl=ColumnTransformer(transformers=[
        ("tnf",StandardScaler(),['Fresh', 'Milk', 'Grocery', 'Frozen', 'Detergents_Paper',
       'Delicatessen'])
],remainder="passthrough")
```

```python
#from sklearn.linear_model import LogisticRegression
#from sklearn.neighbors import KNeighborsClassifier,KNeighborsRegressor
```

```python
#  KNeighborsClassifier
pipe=Pipeline([
    ("preprocessor",cl),("model",KNeighborsClassifier(n_neighbors=5))
])
```

```python
pipe.fit(x_train,y_train)
```

```python
y_pr=pipe.predict(x_test)
```

```python
y_pr
```

```python
print(accuracy_score(y_test,y_pr))
```

```python
#KNeighborsRegressor
pipe1=Pipeline([
    ("preprocessor",cl),("model",KNeighborsRegressor(n_neighbors=5))
])
```

```python
pipe1.fit(x_train,y_train)
```

```python
y_pr1=pipe1.predict(x_test)
y_pr1
```

```python
print(r2_score(y_test,y_pr1))
print(mean_absolute_error(y_test,y_pr1))
print(mean_squared_error(y_test,y_pr1))
```

```python
#from sklearn.tree import DecisionTreeClassifier,DecisionTreeRegressor
```

```python
pipe2=Pipeline([
    ("preprocessor",cl),("model",DecisionTreeClassifier(random_state=42,max_depth=5,criterion="gini"))
])
```

```python
pipe2.fit(x_train,y_train)
```

```python
y_pr2=pipe2.predict(x_test)
y_pr2
```

```python
print(accuracy_score(y_test,y_pr2))
```

```python
pipe3=Pipeline([
    ("preprocessor",cl),("model",LogisticRegression(random_state=42))
])
```

```python
pipe3.fit(x_train,y_train)
```

```python
y_pr3=pipe3.predict(x_test)
y_pr3
```

```python
print(accuracy_score(y_test,y_pr3))
```
